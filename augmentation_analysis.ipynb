```python
import os
import torch
from tensorboard.backend.event_processing import event_accumulator

HISTORY_PATHS = [f"history_v{i}.pt" for i in range(1, 16)]
EVENTS_DIRS = [f"events_v{i}" for i in range(1, 16)]


# ============
# history.pt
# ============
def read_history(history_path):
    print("\n=== HISTORY.PT ===")
    
    if not os.path.exists(history_path):
        print(f"[ERROR] {history_path}")
        return None

    history = torch.load(history_path, map_location="cpu", weights_only=False)

    for key, value in history.items():
        if isinstance(value, list):
            print(f"{key}: {value[-1] if value else 'no data'}")
        else:
            print(f"{key}: {value}")

    return history


# ============
# events.out.tfevents.*
# ============
def read_events(events_dir):
    print("\n=== TENSORBOARD EVENTS ===")

    events_files = [f for f in os.listdir(events_dir) if "tfevents" in f]

    if not events_files:
        print("[ERROR] 'events.out.tfevents.*'")
        return
    
    for fname in events_files:
        print(f"\n>>> File: {fname}")
        ea = event_accumulator.EventAccumulator(os.path.join(events_dir, fname))
        ea.Reload()

        tags = ea.Tags()
        print("Scalars:", tags.get("scalars", []))

        if "scalars" in tags:
            for tag in tags["scalars"]:
                print(f"\n--- {tag} ---")
                values = ea.Scalars(tag)
                for entry in values:
                    print(f"step={entry.step}, value={entry.value}")


if __name__ == "__main__":
    i = 0
    for history_path, events_dir in zip(HISTORY_PATHS, EVENTS_DIRS):
        i += 1
        print(f"\n############# VERSION {i} #############")
        read_history(history_path)
        #read_events(events_dir)
```
